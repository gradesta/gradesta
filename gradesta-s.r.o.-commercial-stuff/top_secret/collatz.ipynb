```python
from typing import List

def numerator(ks: List[int]) -> int:
    """
    Compute the numerator

        2⋅3⁸
      + 3⁷⋅2^{k₀-1}
      + 3⁶⋅2^{k₀+k₁-1}
      + …
      + 3¹⋅2^{k₀+…+k₆-1}
      + 3⁰⋅2^{k₀+…+k₇-1}
      − 2^{k₀+…+k₈-1}

    Parameters
    ----------
    ks : list[int]
        Exactly nine integers [k₀,…,k₈]

    Returns
    -------
    int
        The value of the numerator.

    Notes
    -----
    • All kᵢ must be ≥ 1, otherwise 2^{…−1} becomes a fraction and the function
      will return a float (because Python’s `**` with a negative exponent
      produces one).
    """


    # First term: 2·3⁸
    value = 2 * 3**(len(ks)-1)

    # Running sum Σ_{j=0}^{i} k_j   (but we start with k₀ later)
    prefix_sum = 0

    # Terms with the pattern 3^{8−i} · 2^{(Σ k) − 1},  i = 1 … 8
    for i in range(len(ks)-1):               # i = 0 … 7 → uses k₀ … k₇
        prefix_sum += ks[i]
        exponent = prefix_sum - 1
        power_of_three = 3**(len(ks)-2 - i)  # 3^{8−(i+1)}
        value += power_of_three * (2**exponent)

    # Final subtractive term
    prefix_sum += ks[len(ks)-1]              # add k₈
    value -= 2**(prefix_sum - 1)

    return value
```

```python
def int_to_base(n: int, base: int, *, keep_sign: bool = True) -> str:
    """
    Convert *any* signed integer to a base-`base` string (2 ≤ base ≤ 36).

    - If keep_sign is True, a leading '-' is added for negative n.
    - If keep_sign is False, the sign is ignored (absolute value is used).
    """
    if not (2 <= base <= 36):
        raise ValueError("base must be in the range 2–36")
    # Special-case zero
    if n == 0:
        return "0"
    digits = "0123456789ABCDEFGHIJKLMNOPQRSTUVWXYZ"
    sign = "-" if (n < 0 and keep_sign) else ""
    n = abs(n)
    out = []
    while n:
        n, r = divmod(n, base)
        out.append(digits[int(r)])
    return sign + "".join(reversed(out))



def ltrim(number: int, base: int, digits: int) -> int:
    """
    1. Keep only the last `digits` base-`base` digits of `number`.
    2. Print those digits as a string in that base.
    3. Return the numeric value represented by those digits.

    Examples
    --------
    ltrim(123456, 10, 3)      # prints 456   -> returns 456
    ltrim(255,     16, 2)     # prints FF    -> returns 255
    ltrim(10,      2, 4)      # prints 1010  -> returns 10
    ltrim(-98765,  36, 4)     # prints -1U1T -> returns -8733
    """
    if not (2 <= base <= 36):
        raise ValueError("base must be between 2 and 36 (inclusive).")
    if digits < 1:
        raise ValueError("digits must be a positive integer.")
    cutoff = base ** digits
    abs_rem = abs(number) % cutoff            # numeric remainder
    str_rem = int_to_base(int(abs_rem), base).zfill(digits)
    # Return signed numeric value
    return abs_rem
```

```python
def rtrim(number: int, base: int, digits: int) -> int:
    """
    Return n digits on the left of the number
    """
    return int_to_base(number, base)[:digits]

```

```python
def permutations(k_sum, n):
    """
    Generator that yields all permutations of `n` positive integers that sum to `k_sum`.
    """
    # The sum must be at least n, since each of the n numbers must be at least 1.
    if k_sum < n or n < 1:
        return

    # Base case: If we only need to find one number.
    if n == 1:
        yield [k_sum]
        return

    # Recursive step: Iterate through possible values for the first number.
    # The first number 'i' must be at least 1.
    # The max value for 'i' leaves enough for the other n-1 numbers (at least 1 each).
    for i in range(1, k_sum - (n - 1) + 1):
        for p in permutations(k_sum - i, n - 1):
            yield [i] + p

# Example for positive integers
result_positive = list(permutations(8, 5))
print(result_positive)
```

```python
# Now I want to see the range of the original numerator values for each of these buckets
# First we will create a class which will store the ltrimmed value and a set of numerator values
# Then we will create a dictionary of these classes
# Finally we will populate the dictionary with the ltrimmed values and the numerator values
# Then we will display the dictionary

class Bucket:
    def __init__(self, ltrimmed_value):
        self.ltrimmed_value = ltrimmed_value
        self.numerator = set()
        self.rtrimmed_values = set()
    def add(self, numerator):
        self.numerator.add(numerator)
        self.rtrimmed_values.add(rtrim(numerator, 2, bits))

    def __repr__(self):
        return f"ltrim={int_to_base(self.ltrimmed_value, int(2))} Numerator count {len(self.numerator)}, min={min(self.numerator)}, max={max(self.numerator)}"

# Create a dictionary of buckets
buckets = {}

# Populate the dictionary with the ltrimmed values and the numerator values
bits = 4
ll = 12
col = ll
denominator = 2**col - 3**ll
while denominator < 0:
    col += 1
    denominator = 2**col - 3**ll
print(f"col={col}, ll={ll}")
print(denominator)
for permuatation in permutations(col, ll):
    ltrimmed_value = ltrim(numerator(permuatation), 2, bits)
    if ltrimmed_value not in buckets:
        buckets[ltrimmed_value] = Bucket(ltrimmed_value)
    buckets[ltrimmed_value].add(numerator(permuatation))

# sort the buckets by min of the numerator values then display them
sorted_buckets = sorted(buckets.values(), key=lambda x: min(x.numerator))
for bucket in sorted_buckets:
    print(bucket)

max_numerator = max(max(bucket.numerator) for bucket in sorted_buckets)
# max c is the maximum whole number constant that can be multiplied by the denominator to get a numerator value that is greater than the maximum numerator value
max_c = floor(max_numerator / denominator)
print(f"max c is {max_c}")

# go through each bucket and display the number of numerator values that are greater than the denominator
for c in range(1, max_c+1):
    print(f"c={c}")
    print("-"*100)
    for bucket in sorted_buckets:
        print(f"Bucket {int_to_base(bucket.ltrimmed_value, 2)} has {len([x for x in bucket.numerator if x > denominator*c])}\nnumerator values greater than the denominator times {c} {denominator*c} ltrim value {int_to_base(ltrim(denominator*c, 2, bits), 2)} rtrim value {rtrim(denominator*c, 2, bits)}")
        print("rtrim values", bucket.rtrimmed_values)
```

```python
def print_3_power_difference_table(powerof3=1):
    """
    prints a table of various permutations of 3^powerof3 *2^i - 3^powerof3 *2^j
    for i and j in the range of 0 to powerof3
    """
    for i in range(1000):
        for j in range(1000):
            # only print if there are at least 5 zeros at the end of the base 3 representation
            if int_to_base(3**powerof3 *2**i - 3**powerof3 *2**j, 3).endswith("00000"):
                print(f"3^{powerof3} * 2^{i} - 3^{powerof3} * 2^{j} = {int_to_base(3**powerof3 *2**i - 3**powerof3 *2**j, 3)}")
                print("Number of zeros at the end of the base 3 representation:", int_to_base(3**powerof3 *2**i - 3**powerof3 *2**j, 3).split("1")[-1].split("2")[-1].count("0"))
print_3_power_difference_table(1)
```
